```python
# Lab 3

import pandas as pd
import numpy as np

# Load Dataset
df = pd.read_csv("train.csv")
```

```python
# PROBLEM 1: DATASET INSPECTION


# 1. Display the first 10 rows
print(df.head(10))

# 2. Find the number of rows and columns
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# 3. Identify all column names
print(df.columns.tolist())

# 4. Determine the data type of each column
print(df.dtypes)

# 5. Display descriptive statistics for numerical columns
print(df.describe())
```

```python
# PROBLEM 2: DATA FILTERING


# 1. Display passengers belonging to first class
print(df[df["Pclass"] == 1])

# 2. Find passengers younger than 18
print(df[df["Age"] < 18])

# 3. Find male passengers who survived
print(df[(df["Sex"] == "male") & (df["Survived"] == 1)])

# 4. Find passengers who paid a fare greater than 100
print(df[df["Fare"] > 100])

```

```python
# PROBLEM 3: SORTING AND AGGREGATION


# 1. Display the 10 passengers who paid the highest fares
print(df.sort_values(by="Fare", ascending=False).head(10))

# 2. Calculate the average passenger age
print("Average Passenger Age:")
print(df["Age"].mean())

# 3. Count the number of male and female passengers
print("Number of Male and Female Passengers:")
print(df["Sex"].value_counts())

# 4. Calculate the average fare for each passenger class
print(df.groupby("Pclass")["Fare"].mean())

# 5. Calculate the survival rate for each passenger class
print(df.groupby("Pclass")["Survived"].mean())

```

```python
# FINAL LABORATORY CHALLENGE


# 1. How many passengers are in the dataset?
print("\n1. Total Number of Passengers:")
print(len(df))

# 2. What is the average passenger age?
print("\n2. Average Passenger Age:")
print(df["Age"].mean())

# 3. Which passenger class has the highest number of passengers?
print("\n3. Passenger Count in Each Class:")
class_counts = df["Pclass"].value_counts().sort_index()
print(class_counts)

print("\nPassenger Class With Highest Number of Passengers:")
print(class_counts.idxmax())

# 4. What is the average fare paid by male and female passengers?
print("\n4. Average Fare by Gender:")
print(df.groupby("Sex")["Fare"].mean())

# 5. Which gender has a higher survival rate?
print("\n5. Survival Rate by Gender:")
gender_survival = df.groupby("Sex")["Survived"].mean()
print(gender_survival)

print("\nGender With Higher Survival Rate:")
print(gender_survival.idxmax())

# 6. How does the survival rate differ across passenger classes?
print("\n6. Survival Rate Across Passenger Classes:")
class_survival = df.groupby("Pclass")["Survived"].mean()
print(class_survival)

print("\nPassenger Class With Highest Survival Rate:")
print(class_survival.idxmax())
```
