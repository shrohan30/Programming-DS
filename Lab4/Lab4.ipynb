```python
#Problem 1

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv")

print("First 5 rows")
print(df.head())

print("numerical & Categorical Attribute")
print(df.columns.tolist())
```

```python
#Problem 2

import matplotlib.pyplot as plt

x = df["Pclass"]
y = df["Fare"]

Average = df.groupby("Pclass")["Fare"].mean()

plt.plot(Average.index, Average.values)
#plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#Problem 3

import matplotlib.pyplot as plt

x = df["Age"]
y = df["Fare"]

plt.scatter(df["Age"], df["Fare"])
plt.title("ScatterPlot")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

```python
#Problem 4

import matplotlib.pyplot as plt

gender = df["Sex"].value_counts()

plt.bar(gender.index, gender.values)

plt.title("BarChart")
plt.xlabel("Gender")
plt.ylabel("Values")

plt.show()
```

```python
#Problem 5

import matplotlib.pyplot as plt

Age = df["Age"]

plt.hist(Age, bins = 15, edgecolor="black")

plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

```python
#Problem 6

import matplotlib.pyplot as plt

Passangers = df["Survived"].value_counts()

plt.pie(
     Passangers,
     labels=["Survived", ["Not Survived"]],
     autopct="%1.1f%%"
)

plt.title("PieChart")

plt.show()
```

```python
#Problem 7

import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2)

# Histogram
ax[0].hist(df["Age"], bins=15, edgecolor = "yellow")
ax[0].set_title("Distribution of Passenger Ages")
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Frequency")

# Bar Chart
class_count = df["Pclass"].value_counts().sort_index()

ax[1].bar(class_count.index, class_count.values)
ax[1].set_title("Passengers by Class")
ax[1].set_xlabel("Passenger Class")
ax[1].set_ylabel("Number of Passengers")

plt.tight_layout()
plt.show()
```

```python
#Problem 8

import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# Bar Chart - Gender
gender = df["Sex"].value_counts()

ax[0].bar(gender.index, gender.values)
ax[0].set_title("Passengers by Gender")
ax[0].set_xlabel("Gender")
ax[0].set_ylabel("Number of Passengers")

# Histogram - Fare
ax[1].hist(df["Fare"], bins=10, edgecolor = "black")
ax[1].set_title("Distribution of Passenger Fares")
ax[1].set_xlabel("Fare")
ax[1].set_ylabel("Frequency")

# Pie Chart - Survival
survival = df["Survived"].value_counts()

ax[2].pie(
    survival,
    labels=["Not Survived", "Survived"],
    autopct="%1.1f%%"
)
ax[2].set_title("Survival Proportion")

plt.tight_layout()
plt.show()
```
